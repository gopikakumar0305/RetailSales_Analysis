```python
orders_df['order_date'] = pd.to_datetime(
    orders_df['order_date'],
    dayfirst=True
)
```

```python
orders_df['order_date'].head(5)
```

```python
rfm_data = pd.merge(
    orders_df[['order_id','customer_id','order_date']],
    order_items_df[['order_id','total_price']],
    on='order_id'
)
```

```python
rfm_data.head()
```

```python
snapshot_date = rfm_data['order_date'].max() + pd.Timedelta(days=1)
print(snapshot_date)
```

```python
rfm = rfm_data.groupby('customer_id').agg({
    'order_date': lambda x: (snapshot_date - x.max()).days,
    'order_id': 'nunique',
    'total_price': 'sum'
}).reset_index()
```

```python
rfm.columns = [
    'customer_id',
    'Recency',
    'Frequency',
    'Monetary'
]
```
